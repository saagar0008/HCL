```python
import os
import pandas as pd

file = "train.csv"

if os.path.exists(file):
    print("CSV file exists!")

    try:
        df = pd.read_csv(file)
        print("CSV file loaded successfully!")
        print(df.head())
    except Exception as e:
        print("Error reading CSV file:", e)

else:
    print("CSV file does not exist!")
```

```python
survival = df["Survived"].value_counts()

print("Survivors:", survival[1])
print("Non-Survivors:", survival[0])
```

```python
print("Passengers by Sex:")
print(df["Sex"].value_counts())

print("\nPassengers by Class:")
print(df["Pclass"].value_counts())
```

```python
age = df["Age"].dropna()
fare = df["Fare"].dropna()

print("Average Age:", age.mean())
print("Average Fare:", fare.mean())
```

```python
print("Survival by Sex:")
print(pd.crosstab(df["Sex"], df["Survived"]))
```

```python
print("Missing Age:", df["Age"].isnull().sum())

print("Missing Cabin:", df["Cabin"].isnull().sum())

print("Missing Embarked:", df["Embarked"].isnull().sum())
```

```python
top_fares = df.sort_values(
    by="Fare", ascending=False
)

print(top_fares[["Name", "Fare"]].head(10))
```

```python
import matplotlib.pyplot as plt

df["Survived"].value_counts().plot(kind="bar")

plt.title("Survival Count")
plt.xlabel("Survived (0=No, 1=Yes)")
plt.ylabel("Passengers")

plt.show()
```

```python
plt.hist(df["Age"].dropna(), bins=20)

plt.title("Passenger Age Distribution")
plt.xlabel("Age")
plt.ylabel("Passengers")

plt.show()
```
